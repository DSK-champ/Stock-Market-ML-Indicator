```python
from pathlib import Path
import numpy as np
import pandas as pd

DATA_PATH   = Path("../data/raw/BTCUSDT_5m.csv")
OUTPUT_PATH = Path("../data/processed/BTCUSDT_5m_clean.csv")

df = pd.read_csv(DATA_PATH, parse_dates=["open_time"])

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
print(f"Data types : {df.dtypes}")
```

```python
print("Dataset information : ")
df.info()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate timestamps:", df["open_time"].duplicated().sum())

print("\nDate range:")
print("Start:", df["open_time"].min())
print("End:  ", df["open_time"].max())

df.describe().T
```

```python
required_columns = [
    "open_time",
    "open", "high", "low", "close",
    "volume", "quote_volume",
    "number_of_trades", 
    "taker_buy_base_volume",
    "taker_buy_quote_volume"
]

missing_columns = set(required_columns) - set(df.columns)

if missing_columns:
    raise ValueError(f"Missing required columns : {missing_columns}")

print("All required columns are present")
```

```python
numeric_columns = [
    "open", "high", "low", "close",
    "volume", "quote_volume", "number_of_trades",
    "taker_buy_base_volume", "taker_buy_quote_volume"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    
df["open_time"] = pd.to_datetime(df["open_time"], utc=True, errors="coerce")

print("Missing values after conversion:")
display(df.isna().sum())

print("\nInfinite values in numeric columns:")
display(
    pd.Series({
        col: np.isinf(df[col].dropna()).sum()
        for col in numeric_columns
    }, name="infinite_values")
)
```

```python
before = len(df)

df = df.dropna(subset=required_columns)
df = df.drop_duplicates(subset=["open_time"], keep="last")

print("Rows before cleaning:", before)
print("Rows after  cleaning:", len(df))
print("Rows removed:", before - len(df))
```

```python
invalid_price = (
    (df["open"]  <= 0) |
    (df["high"]  <= 0) |
    (df["low" ]  <= 0) |
    (df["close"] <= 0)
)

invalid_ohlc = (
    (df["high"] < df[["open", "close", "low" ]].max(axis=1)) |
    (df["low"]  > df[["open", "close", "high"]].min(axis=1))
)

invalid_volume = (
    (df["volume"] < 0) |
    (df["quote_volume"] < 0) |
    (df["number_of_trades"] < 0) |
    (df["taker_buy_base_volume"] < 0) |
    (df["taker_buy_quote_volume"] < 0)
)

print("Invalid prices                : ", invalid_price.sum())
print("Invalid OHLC candles          : ", invalid_ohlc.sum())
print("Negative volume/trade values  : ", invalid_volume.sum())
```

```python
df = df.sort_values("open_time").reset_index(drop=True)

time_diff = df["open_time"].diff()

print("Chronologically sorted          : ", df["open_time"].is_monotonic_increasing)
print("Duplicate timestamps            : ", df["open_time"].duplicated().sum())
print("Intervals other than 5 minutes  : ", (time_diff.dropna() != pd.Timedelta(minutes=5)).sum())
print("Missing values                  : ", df.isna().sum().sum())
```

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(OUTPUT_PATH, index=False)

print("Cleaned dataset saved to:", OUTPUT_PATH.resolve())
print("Final shape : ", df.shape)
print("Start       : ", df["open_time"].min())
print("End         : ", df["open_time"].max())
```
